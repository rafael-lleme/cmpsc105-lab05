## Plot 1

**Scenario**: A researcher wants to show that Gentoo penguins are significantly heavier than other species, but their plot is visually overwhelming and exaggerates the differences.

```python
penguins = sns.load_dataset("penguins")

ax = sns.barplot(
    data=penguins,
    x="species",
    y="body_mass_g",
    color="steelblue"
)

ax.set_ylim(0, 5500)
ax.set_xlabel("Species")
ax.set_ylabel("Average Body Mass (g)")
plt.title("Average Penguin Body Mass by Species")

plt.show()
```

Issues with the plot: The y axis starts at 3000 instead of 0, what visually exagerates the weight difference between species. The colors also are really bright and different between eachother, without representing a single addition information, making the grafic visually poluted. Additionaly, the name of the y axis (body_mass_g) isn't as clear as it should for the reader, and it definitely could indicate that the values represents body mass in grams in a better way. The title could also make it clear that the bars represent the body weight's averages of each species.


How you changed the plot and why: I altered the y axis to start at 0, making so that the difference between species are represented in a proporcional way and without visual polution. Additionaly, I also changed the different bright color to a singular color, since the color don't represent any additional variables. Lastly, I altered the title and the axis's names to make the grafic clearer and easier to interpret.

## Plot 2

**Scenario**: A student plotted the carat size versus price for 54,000 diamonds to see where the bulk of the market sits, but the result is just a solid block of ink.

```python
diamonds = sns.load_dataset("diamonds")

sns.scatterplot(
    data=diamonds,
    x="carat",
    y="price",
    color="blue",
    s=10,
    alpha=0.15
)

plt.xlabel("Carat")
plt.ylabel("Price ($)")
plt.title("Diamond Price vs. Carat")
plt.show()
```

Issues with the plot: The graph has too many overlapping data points, making it hard to identify where the majority of the diamonds are concentrated. The size of the data points (s=100) is too big for a database with approximately 54,000 observations. Additionally, the data points are completely opaque, making them overlap with each other. The names of the axes could also be more descriptive, principally indicating that the price is in USD ($).

How you changed the plot and why: I made the data points significantly smaller and also added transparency. That way, areas with less observations are clearer, while regions with a lot of overlapping data points are more visually apparent, making it easier to identify where the majority of the diamonds are concentrated. I also altered the names of the axes to make the units clearer.

## Plot 3

The aviation board wants to see passenger growth month-over-month for each year in the 1950s, but the lines are indistinguishable.

```python
flights = sns.load_dataset("flights")

sns.lineplot(data=flights, x="month", y="passengers", units="year", estimator=None, color="black")
plt.title("Passengers over Time")
plt.show()
```

Issues with the plot: All the lines have the same color, making it impossible to distinguish the different years, also, there's no text indicating which line represents each year. Additionally, the title "Passengers over time" is generic and doesn't make it clear the we are comparing the years. Lastly the name of the axes (month and passengers) could be more descriptive, also, because there is too many lines close/crossing each other, using only black lines visually confusing.


How you changed the plot and why: I would change the colors of the lines based on each year, in order to distinguish the years from each other. I also would add a legend to identify which line represents each year. Lastly, I would change the title and the names of the axes to make the information represented by the graph clearer.


## Plot 4

**Scenario**: A restaurant manager receives this chart showing the relationship between total bills and tips, split by smoking preference. 





```python
tips = sns.load_dataset("tips")

sns.lmplot(data=tips, x="total_bill", y="tip", hue="smoker")
plt.show()
```

Issues with the plot: The graph doesn't have a title, making it harder to immediately understand what is being analyzed. The names of the axes are also not very descriptive, and don't indicate that both the total bill and tip are measured in USD ($). The name of the legend "smoker" could also be more descriptive about what the groups represent. Additionally, the shaded areas around the lines can be confusing for someone who doesn't know what they represent.


How you changed the plot and why: I would add a title to make it clearer what the graph is analyzing, and would also change the names of the axes to indicate that the values are measured in USD ($). I would change the name of the legend to make it clearer that the groups represent smoking preference. Finally, I would remove the shaded areas around the lines to make the graph less visually distracting and easier to understand.
